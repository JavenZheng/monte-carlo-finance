## 一、先说这门课要干什么

你要读的那本书有六百页，讲怎么用电脑算金融里的不确定性。这门预备课要做的是另一件事：在你翻开它之前，先让你手里有一样东西。

不是公式，不是编程。是一个你遇到没见过的问题时能用的动作。

所以第一节我们不碰电脑，也不写代码。我们用六张便利贴，把那个动作做出来一次。做完你会发现，你早就理解蒙特卡洛了，只是不知道它叫这个名字。

## 二、你的账号，连着六条没爆

假设你在经营一个自己的账号。发了一段时间，你心里有个模糊的判断：大概六条里能爆一条。

这个判断不精确，但它是你手上唯一的东西。

这周你连发六条。一条都没爆。

于是要么是你那个"六条爆一条"的感觉错了，要么这六条内容确实有问题，要么就是运气。这三种解释在你脑子里长得一模一样，你都分不开。你大概会做那件最常见的事：改。换选题、换封面、换发布时间，因为六条没爆让你坐不住了。

我们把这件事变成一个能回答的问题：

> 如果"六条爆一条"是真的，那么连发六条一条都没爆，这种事多久发生一次？

注意这个问题问的不是"我这次为什么没爆"，那没法回答。问的是：在你说的那个规律之下，这种结果有多常见。

这个问题，用便利贴就能做。

## 三、把世界搭出来

拿六张便利贴，大小尽量一样。在其中一张上画一个点，其余五张留白。全部揉成团，放进杯子或者摊在桌上。

现在：

- 画点的那张，代表"这条爆了"。
- 空白，代表"没爆"。
- 你抽一张，看一眼，**放回去**，再抽。抽六次，就是"发六条"。
- 这六次里如果一次都没抽到点，就在本子上记一个"正"字。抽到过点的，什么都不记。

抽满六次算一轮。做十轮，你会得到"十个里面有几个'六条全空'"。

**这就是蒙特卡洛。** 你已经做完了，只是还不知道它叫什么：你手上没有关于这个问题的公式，于是你造了一个小小的世界，让它按你猜的规律一遍遍发生，然后数结果。

那句"放回去"很重要，别省。每次抽完不放回，第二次的概率就变了——你需要让每一次抽都是独立的一发，就像你每条视频各自面对一次推荐池的抽签。

## 四、你手工做出来的答案，会不太好看

我做了一次这件事，也把很多人做这件事的结果统计过。真实的规律是这样的：

**六条全空，大约每三次里就有一次。**（准确说是 33.49%。）

也就是说，如果"六条爆一条"的感觉是对的，你连着六条不发只是运气差，这件事三分之一的概率会发生。六条全空根本不该被当成信号。

但你自己做十轮，不会得到 33%。十轮太少：有一半的人做出来会落在 20% 到 40% 之间，只有约四分之一的人运气好、估出来的东西在 33.5% 上下五个点以内。更有约 2% 的人十轮一次都没抽到"全空"，于是他会得出"六条全空基本不可能发生"的错误结论——正好和他的亲身经历相反。

这是第一件值得记住的事：**手工做十轮，你会得出一个和你邻座不一样的答案。** 这不是你算错了，是十遍本来就太少。

于是第二件事就来了：**这个办法的全部威力，取决于你能演多少遍。** 而演一万遍，人做不到，电脑随手就做完了。

我在自己电脑上替你演了一百万遍，得到 33.57%。真值是 33.49%。

这就是这门课要交给你的那台机器。你以后不再靠十轮的运气，你演一百万轮。

## 五、你今晚就能用的三样东西

这一节把上面做的东西换成你能拿走的具体用法。这三件事和便利贴无关，但都是同一招。

**第一个问题：我这个连续失败，算不算异常。**

你手上有个大致的成功率，然后你连着失败了 n 次，你在纠结这是不是说明"方法不行"。你不需要任何公式，只需要问：如果那个成功率是真的，"连续 n 次全空"多久出现一次？

按六分之一的爆率算，几种长度的连败有多常见：

| 你连发几条 | 一条都没爆的概率 | 大概多久遇上一次 |
|---|---|---|
| 3 条 | 57.9% | 十次里有六次 |
| 6 条 | 33.5% | 三次里一次 |
| 13 条 | 9.3% | 十次里一次 |
| 20 条 | 2.6% | 三十八次里一次 |

看到那条曲线了吗：三条不爆是常事，六条不爆仍常事，要连到二十条才真正值得怀疑你的那个"六分之一"。

**所以：你打算在第几条之后开始改打法？这本身是个可以自己定的阈值。** 如果你把"改打法"的触发点设在六条，那你每隔几次就会改一次，你的风格永远长不起来。很多人不是被失败打败的，是被一个本来该出现的正常波动打败的。

**第二个问题：我要做到多少把握，得发多少条。**

反过来算。你想"这批里至少爆一条的把握到九成"，那得发多少条？按六分之一算，答案是十三条（此时 90.7%）。

这个数字是可以直接拿去做计划的：你决定做一期系列，想让"至少出一条爆款"的把握够高，那你预算的条数应该是按把握算出来的，不是按心情算出来的。

**第三个问题，也是最重要的一个：这套办法什么时候该用。**

不是所有问题都适合。适合的问题是这三种形状：

- 你知道规律，不知道结果。（知道爆率，不知道这次会不会全空。）
- 结果由很多步串起来，一步算不出来。（期权价格来自一路上百万种走法。）
- 你能造出一个"假世界"，而且它和你相信的真实规律大体一致。

不适合的是：你不知道规律，也没法验证它。比如"我这条视频能不能爆"，如果你连爆率是多少都猜不出，那便利贴演一百万遍也是在演一个你随手编的世界，答案没有任何意义。这一点下一节展开。

## 六、一件事要提醒你

回到第三节那个杯子。

你有没有想过一件事：你演出来的所有结果，说的都是**这个杯子里的世界**，不是你账号的世界。

如果你那个"六条爆一条"其实感觉错了，真实情况是二十条爆一条，那么"连发六条全空"的概率就不是 33.5%，而是 73.5%。你的便利贴会一百万次地、精确地告诉你一个错的答案。

再比如你便利贴撕得大小不一，那张画了点的偏偏比较大，你更容易摸到它——你演出来的爆率会比你以为的高。

这就是这门课以后反复要处理的东西：**演得多，治的是你手气不好；演得多治不了你把世界搭错了。** 前者后面有大量技术去改善，后者只能靠你不断回头检查：我这个规律到底是不是真的。

现在你只需要记住一句话：**蒙特卡洛给你的答案，最准也只能准到你那个假世界本身。**

## 七、练习

**一。** 你在一家公司做销售。你估摸着每个新客户有五分之一的机会签单。这个季度你跟了五个客户，全都没签。用这一节课的思路，说说这有多不寻常。你想动手验证的话，便利贴要准备几张？

**二。** 上一题的"全都没签"如果让你决定"以后不做陌生了，全部转老客户"，那你是在第几条证据上做决定的？按你课上算出来的那张表，这个决定合理吗？

**三。** 你朋友跟你说："我做了十轮实验，六条全空只发生过一次，所以这世界上六条全空的概率就是 10%。" 用第四节的哪个事实能反驳他？

**四。** 下面每个问题，判断"适合用蒙特卡洛"、"不适合"、或者"取决于你还不知道的东西"：
（a）我下个月的花销会不会超支；
（b）我该不该接受这份新工作；
（c）一支含三次分红的股票期权值多少钱；
（d）明天会不会下雨。

### 答案

**一。** 每张纸代表一个客户，五张空白、一张画点，抽五次记一次结果。真值是 (4/5)^5 = 32.8%，也就是说五个客户全签不到，本来就有三分之一的概率会发生，属于常见结果。

**二。** 32.8% 的事情，你几乎每个季度都会碰上。按第五节的表，你的"改打法阈值"应该设在更长的连败之后（比如连续十几个客户全空，那时概率才掉到 10% 以下）。在这之前改，你会反复改，什么都积累不下来。

**三。** 第四节那个统计：做十轮的人，只有一半落在 20%–40%，估准的只有约四分之一。他那"10%"不是发现，是他那十轮的运气。要压住这个散布，得把轮数加大——这就是我们下一课要用电脑的原因。

**四。**
（a）适合，前提是你已经能大致说出每类开销怎么波动。这恰好是这门课会做的事。
（b）不适合。这不是"知道规律、不知道结果"，你缺的是信息和对自己的了解，演多少遍都没用。承认这一点很重要：不是所有难题都能算。
（c）适合，而且这就是那本书的主要内容：一条路径算不出来，但路径可以由很多小步串成，每一步的规律已知，于是能演。
（d）取决于你有没有一个可信的天气模型。有模型，蒙特卡洛就有意义；只是"看云猜"，演一百万遍也只是把你的猜测放大。

**可选验算（可跳过）。** 上面答案一里印的那个 32.8%，用手就能核：每条签不到的概率是 4/5，五个客户相互独立。这一格不属于"必须做"的部分，你不想碰电脑可以直接往下读。

```python
# 可选验算（跳过不影响这一课）：练习一的真值
print(f"(4/5) ** 5 = {(4 / 5) ** 5:.5f} ≈ {(4 / 5) ** 5 * 100:.1f}%")
```

## 进阶框（可跳过，不影响后面）

这一节的数字其实不用演也能算，因为"六条全空"这种问题足够简单：每一条不爆的概率是 5/6，六条相互独立，所以全空是

$$\left(\frac{5}{6}\right)^{6}=33.49\%.$$

那为什么还要演？三个理由，它们会贯穿整门课。

一是演一次你就得到了对规律的**手感**，而公式只给你一个数。第三节那个"放回去"为什么必须做，你演过一遍才真的懂。

二是这类能手算的问题，是检验你的模拟程序写没写错的唯一办法。**写模拟最大的风险是程序悄悄错了，而答案看起来完全合理。** 所以课程里凡是可以手算的题，我们都会先算再演，两个数对上了才继续。

三是你很快会碰到根本手算不了的问题——路径依赖、多次分红、提前行权。那时"演"不是替代公式的笨办法，而是唯一的办法。

至于"演很多遍为什么会稳定下来"，它的名字叫大数定律；而"演十遍为什么会飘成那样、要演多少遍才够稳"，我们会在第 07 课给你一个能量化毛边的式子。这两件事现在只需要知道名字。

**可选验算（可跳过）。** 上面那个式子写在这里像一句承诺，这一格把它按出来，看看是不是课文印的 33.49%。这一课不用电脑也照样成立，跳过它什么都不影响。

```python
# 可选验算（跳过不影响这一课）：六条全空的真值
print(f"(5/6) ** 6 = {(5 / 6) ** 6:.6f} ≈ {(5 / 6) ** 6 * 100:.2f}%")
```

## 备忘卡

本课出现的说法，正式名字在这里。

**一轮**，正式叫**一次实现 / 一条样本路径**（realization / sample path）：把整个随机过程完整走一遍，这里就是"抽六次 = 发六条"。**重复很多轮**叫**模拟次数**（number of simulations）。

**你造的那个小世界**，正式叫**模型**（model）。你的"六条爆一条"是模型里的**参数**（parameter）。

**数出来的比例**，正式叫**估计量**（estimator）：它是你从有限次演练里得到的那个数。

**独立**：每一次抽签互不影响，这是"抽完必须放回"背后的要求。**蒙特卡洛**（Monte Carlo）：靠大量重复随机试验来回答算不出来的问题的一整套方法。
